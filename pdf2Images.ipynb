## Importations

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import pytesseract
import random
import os
from pdf2image import convert_from_path
```

# Unix

#### Pour convertir plusieurs pdfs en images

```python
pdf_folder = '../../PDFs/data' 
image_folder = './IMAGES'  # où les images seront enregistrées

# creer le dossier IMAGES s'il n'existe pas
if not os.path.exists(image_folder):
    os.makedirs(image_folder)

# compteur pour nommer les fichiers
image_counter = 1

# pour tous les fichiers PDF dans le dossier PDFs
for pdf_file in sorted(os.listdir(pdf_folder)):
    if pdf_file.endswith('.pdf'):
        pdf_path = os.path.join(pdf_folder, pdf_file)
        
        # convertir chaque PDF en une liste d'images (une par page)
        images = convert_from_path(pdf_path, dpi=600)

        # si le PDF contient plusieurs pages, ajouter un suffixe pour chaque image
        suffix = ""
        for i, image in enumerate(images):
            # si plusieurs pages dans le PDF, ajouter un suffixe pour les nommer correctement
            if i > 0:
                suffix = chr(97 + i)  # 'b', 'c', etc. (ascii 97 = 'a')

            # nommer et enregistrer l'image
            image_name = f"{image_counter}{suffix}.png"
            image_path = os.path.join(image_folder, image_name)
            image.save(image_path, 'PNG')

        image_counter += 1
```

#### Pour convertir un seul pdf en image

```python
import os
from pdf2image import convert_from_path

pdf_path = '../../PDFs/data/my_pdf.pdf' 
image_folder = './IMAGES' 

if not os.path.exists(image_folder):
    os.makedirs(image_folder)

image_counter = 1

images = convert_from_path(pdf_path, dpi=600)

suffix = ""
for i, image in enumerate(images):
    if i > 0:
        suffix = chr(97 + i) 

    image_name = f"{image_counter}{suffix}.png"
    image_path = os.path.join(image_folder, image_name)
    image.save(image_path, 'PNG')

print(f"Images sauvegardées dans le dossier '{image_folder}'.")
```
